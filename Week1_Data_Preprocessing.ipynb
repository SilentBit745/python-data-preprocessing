```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Create the dataset
data = {
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Aman", "Neha"],
    "Hours_Studied": [2, 5, np.nan, 7, 2, 4],
    "Attendance": [60, 85, 70, np.nan, 60, 90],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Marks": [45, 85, 60, 90, 45, np.nan]
}

df = pd.DataFrame(data)

print("1. Original Dataset")
display(df)

# 2. Check missing values
print("2. Missing Values")
print(df.isnull().sum())

# 3. Remove duplicate rows
df = df.drop_duplicates()

# 4. Fill missing numerical values with averages
columns = ["Hours_Studied", "Attendance", "Marks"]

for column in columns:
    df[column] = df[column].fillna(df[column].mean())

print("3. Cleaned Dataset")
display(df)

print("Missing values after cleaning:")
print(df.isnull().sum())

# 5. Encode categorical data
df = pd.get_dummies(df, columns=["Gender"], dtype=int)

# 6. Normalize numerical columns
scaler = MinMaxScaler()
df[columns] = scaler.fit_transform(df[columns])

print("4. Final Processed Dataset")
display(df)

# 7. Explore the dataset
print("5. Dataset Size:")
print(df.shape)

print("6. Summary Statistics:")
display(df.describe())

# 8. Save the final dataset
df.to_csv("cleaned_student_data.csv", index=False)

print("7. Dataset saved successfully!")
```
